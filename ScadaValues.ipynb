```python
import pandas as pd
import numpy as np
from google.colab import files

# Generate 1,000-row dataset that produced your previous successful results
timesteps = pd.date_range(start="2026-05-01", periods=1000, freq="10min")
np.random.seed(42)
base_ws = np.random.normal(8.0, 1.5, size=1000)
base_wd = np.random.normal(270.0, 15.0, size=1000) % 360

mock_df = pd.DataFrame({
    "timestamp": timesteps, "wind_speed": base_ws, "wind_direction": base_wd,
    "power_output": base_ws * 480 + np.random.normal(0, 50, 1000),
    "rotor_speed": base_ws * 1.5 + np.random.normal(0, 0.2, 1000),
    "pitch_angle": np.where(base_ws > 12, (base_ws - 12) * 2, 0.1),
    "nacelle_temp": 42.0 + np.random.normal(0, 2, 1000),
    "vibration": 1.8 + np.random.normal(0, 0.3, 1000)
})

# Inject the exact structural anomalies
mock_df.loc[50, ['power_output', 'rotor_speed', 'pitch_angle']] = [0.0, 0.0, 90.0]
mock_df.loc[150, 'vibration'] = 9.8

# Save and trigger download to your local computer
mock_df.to_csv("scada_data.csv", index=False)
files.download("scada_data.csv")
print("✅'scada_data.csv' has been downloaded to your computer!")
```
